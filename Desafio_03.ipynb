```python
#Escala de Automonitoramento Emocional. Avaliação dos últimos 14 dias.#
questoes = [
    "a) Não tive interesse ou ânimo para fazer as coisas",
    "b) Reclamei mais do que agi",
    "c) Percebi irregularidade no sono e no apetite",
    "d) Tive vontade de chorar por qualquer motivo",
    "e) Tive dificuldade para me expressar",
    "f) Tive dificuldade para agir ou tomar decisões",
    "g) Senti angústia e preocupaçõ excessiva",
    "h) Senti vergonha ou culpa por minhas escolhas e atual situação",
    "i) Perdi a confiança em mim e nas pessoas",
    "J) Senti vontade de desistir, fugir e desaparecer"
]

pontuacao = 0

for questoes in questoes:
  print("\n" + questoes)
  print("1 - Nunca")
  print("2 - Poucos Dias")
  print("3 - Metade dos Dias")
  print("4 - Quase Todos os Dias")
  resposta = input("Informe Sua Repsosta (1 - 4): ")

  while resposta not in ["1", "2", "3", "4"]:
    print("RESPOSTA INVÁLIDA. TENTE NOVAMENTE!: ")
    resposta = input("Informe Sua Resposta  (1 - 4): ")

  resposta = int(resposta)

  if resposta == 1:
    pontuacao += 2
  elif resposta == 2:
    pontuacao += 1
  elif resposta == 3:
    pontuacao -= 1
  elif resposta == 4:
    pontuacao -= 2

print("Pontução Final:", pontuacao)

if pontuacao >= 6:
    print("Equilíbrio Emocional: Você Apresenta Boa Estabilidade Emocional!")

elif pontuacao >= 4:
    print("Atenção: Há Sinais Leves de Desgaste Emocional")

elif pontuacao >= 2:
    print("Sinal de Alerta: Você Pode Estar Com Sofrimento Emocional Mais Presente")

else:
    print("ALTO RISCO EMOCIONAL: É IMPORTANTE PROCURAR AJUDA PROFISSIONAL")

print("ESTE TESTE NÃO SUBSTITUI UM PROFISSIONAL DE SAÚDE!")
```
