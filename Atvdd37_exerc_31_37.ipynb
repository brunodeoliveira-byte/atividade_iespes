```python
#37.Desenvolva um programa que classifique um cliente para desconto com base no valor da compra:#
#● Até R 100→semdesconto●AtéR 500 → 5% ● Até R 1000→10●AcimadeR 1000 → 15%#

valor = float(input("Informe o valor da compra: R$ "))

if valor > 1000:
  taxa_desconto = 15
elif valor > 500:
  taxa_desconto = 10
elif valor > 100:
  taxa_desconto = 5
else:
 taxa_desconto = 0

desconto = valor * (taxa_desconto/100)
valor_final = valor - desconto
print(f"Sua compra recebeu {taxa_desconto}% de desconto. O valor final da sua compra é R$ {valor_final}")
```

```python
#36. Crie um programa que simule um sistema de senha com 3 tentativas:#
#● Se acertar → acesso liberado ● Após 3 erros → acesso bloqueado#
usuario_correto = "Bruno"
senha_correta = "1234"

tentativa = 0
max_tentativa = 3
while tentativa < max_tentativa:
  usuario = input("Informe seu login: ")
  senha = input("Informe sua senha: ")

  if usuario == usuario_correto and senha == senha_correta:
    print("Acesso Liberado")
    break

  else:
    tentativa += 1
    print(f"Usuário ou Senha Incorretos. Número de Tentativa: {tentativa} de {max_tentativa}")
    if tentativa == max_tentativa:
      print("Acesso Bloqueado")

```

```python
#35.Desenvolva um programa que verifique se um ano é bissexto:#
#● Divisível por 4 → possível bissexto ● Divisível por 100 → não é ● Divisível por 400 → é bissexto#

ano = int(input("Digite um ano: "))

if ano % 400 == 0:
          print(f"O ano {ano} é considerado bissexto")
elif ano % 100 == 0:
          print(f"O ano {ano} não é considerado bissexto")
elif ano % 4 == 0:
          print(f"O ano {ano} é considerado bissexto")
else:
          print("O ano Não é bissexto")
```

```python
#34. Crie um programa que classifique um funcionário com base no desempenho (nota de 0 a 10):#
# ● Excelente ● Bom ● Regular ● Ruim#

funcionario = input("Informe o nome do Funcionário: ")
desempenho = float(input("Informe a Nota de Desempenho do Funcionário: "))

if desempenho >= 9:
  resultado = "Excelente"
elif desempenho >= 7:
  resultado = "Bom"
elif desempenho >= 5:
  resultado = "Regular"
else:
  resultado = "Ruim"

print(f"A nota de desempenho que o funcionário {funcionario} obteve foi de {desempenho}")
print(f"Esse resultado é considerado {resultado}")
```
