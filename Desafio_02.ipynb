```python
#SISTEMA DE CONTROLE DE ESTOQUE#

material = input("Informe o Nome do Material: ")

print ("\nDefinir Controle de Conferência:")
print ("Caixa: cx")
print ("Pacote: pct")
print ("Unidade: und")
tipo_controle = input("Controle de Conferência Definido: ")

situacao_inicial = int(input("Situação do Estoque: "))
entrada = int(input("Quantidade que Entrou: "))
saida = int(input("Material Retirado Para Abastecimento: "))
atualizacao_estoque = ((situacao_inicial+entrada)-saida)
print("Material:", material)
print("A situação atualizada do estoque é:", atualizacao_estoque, tipo_controle)
```
