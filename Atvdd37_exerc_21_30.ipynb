```python
#30. Crie um programa que calcule o valor de uma conta de energia com base no consumo:#
#● Até 100 kWh → tarifa básica ● Até 200 kWh → tarifa média ● Acima de 200 kWh → tarifa alta#

consumo = float(input("Informe o valor do consumo: "))
t_basica = 15
t_media = 25
t_alta = 40

if consumo <= 100:
  situacao = consumo * t_basica
  print(f"O seu consumo de energia enquadra-se na tarifa básica. O pagamento será R$ {situacao}")
elif consumo <= 200:
  situacao = consumo * t_media
  print(f"O seu consumo de energia enquadra-se na tarifa média. O pagamento será de R$ {situacao}")
else:
  situacao = consumo *t_alta
  print("O seu consumo de energia enquadra-se na tarifa alta. O pagamento será R$ {situacao}")
```

```python
#29. Desenvolva um programa que calcule o valor de uma corrida de aplicativo com base na distância:#
#● Até 5 km → R 10●Até10km→R 20 ● Até 20 km → R 30●Acimade20km→R 50#

distancia = float(input("Informe o valor da distância que será percorrida: "))


if distancia < 5:
  valor = 10
elif distancia <= 10:
  valor = 20
elif distancia <= 20:
  valor = 30
else:
  valor = 50

print(f"Serviço Aceito. O pagamento será de R$ {valor}")
```

```python
#23. Faça um programa que calcule o preço de um produto com base na forma de pagamento:#
#a. À vista → desconto b. Parcelado → valor normal#

produto = input("informe o nome do produto: ")
preco = float(input("Informe o preço do produto: "))
quantidade = int(input("Informe a quantidade a ser comprada: "))

preco_total = preco * quantidade

print("Forma de pagamento:")
print("1 - À Vista com Desconto")
print("2 - Parcelado")
print("3 - Preço Normal")

opcao = int(input("Escolha uma opcao: "))

if opcao == 1:
  desconto = float(input("Taxa de Desconto(%): "))
  valor_final = preco_total - (preco_total * desconto / 100)
  print(f"Valor a Pagar: R$ {valor_final}")

elif opcao == 2:
  parcelas = int(input("Informe o número de Parcelas: "))
  valor_parcelas = preco_total / parcelas
  print(f"O pagamento será efeutado em {parcelas}x de R$ {valor_parcelas}")

elif opcao == 3:
  print(f"Valor Total Sem Desconto: R$ {preco_total}")

else:
  print("opção Invalida!")


```
