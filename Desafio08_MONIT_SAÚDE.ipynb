```python
# SISTEMA DE MONITORAMENTO DE SAÚDE #

print("\n #====== SISTEMA DE MONITORAMENTO DA SAÚDE ======#")

pontuacao = 0

print("\n ====== DADOS DO USUÁRIO ======")

sexo = input("Informe seu sexo (masculino/feminino): ").lower()
idade = int(input("Informe sua idade: "))

print("\n ====== ALIMENTAÇÃO ======")

refeicoes = input("Realiza as 3 refeições principais (sim/não): ").lower()
if refeicoes == "sim":
    pontuacao += 2

horarios = input("Alimenta-se em horários regulares (sim/não): ").lower()
if horarios == "sim":
    pontuacao += 2

fruta = input("Consome alguma fruta ao menos uma vez na semana (sim/não): ").lower()
if fruta == "sim":
    pontuacao += 2

lanche = input("Substitui uma das 3 refeições por lanche durante a semana (sim/não): ").lower()
if lanche == "sim":
    pontuacao -= 2

bebida = input("Consome refrigerante com maior frequência do que suco (sim/não): ").lower()
if bebida == "sim":
    pontuacao -= 2

comida = input("Consome com maior frequência alimentos enlatados e processados (sim/não): ").lower()
if comida == "sim":
    pontuacao -= 2

caldo = input("Consome sopa, cozido ou canja ao menos 1 vez na semana (sim/não): ").lower()
if caldo == "sim":
    pontuacao += 2

perda_apetite = input("Notou perda de apetite por motivos de preocupação, estresse ou tristeza (sim/não): ").lower()
if perda_apetite == "sim":
    pontuacao -= 2

comer_excesso = input("Notou excesso de apetite por motivos de preocupação, estresse ou tristeza (sim/não): ").lower()
if comer_excesso == "sim":
    pontuacao -= 2

print("\n ====== SONO ======")

dorme_cedo = input("Costuma dormir antes das 23 horas (sim/não): ").lower()
if dorme_cedo == "sim":
    pontuacao += 2

dificuldade = input("Tem dificuldade para dormir (sim/não): ").lower()
if dificuldade == "sim":
    pontuacao -= 2

sono_interrompido = input("Costuma acordar durante a noite (sim/não): ").lower()
if sono_interrompido == "sim":
    pontuacao -= 2

insonia = input("É frequente a ocorrência de insônia (sim/não): ").lower()
if insonia == "sim":
    pontuacao -= 2

ronco = input("É frequente você roncar durante o sono (sim/não): ").lower()
if ronco == "sim":
    pontuacao -= 2

sonolencia = input("É comum você sentir cansaço e sonolência em excesso durante o dia (sim/não): ").lower()
if sonolencia == "sim":
    pontuacao -= 2

descanso = input("Costuma ter um horário para descansar durante o dia (sim/não): ").lower()
if descanso == "sim":
    pontuacao += 2

horas_sono = int(input("Quantas horas de sono você dorme por noite (h): "))
if horas_sono >= 7:
    pontuacao += 2
elif horas_sono >= 5:
    pontuacao += 0
else:
    pontuacao -= 2

perda_sono = input("Notou perda de sono por motivos de preocupação, estresse ou tristeza (sim/não): ").lower()
if perda_sono == "sim":
    pontuacao -= 2

print("\n ====== ATIVIDADE FÍSICA ======")

tempo = input("Você tem tempo para praticar atividade física (sim/não): ").lower()
if tempo == "sim":
    pontuacao += 2

frequencia = input("Costuma praticar alguma atividade física regularmente (sim/não): ").lower()
if frequencia == "sim":
    pontuacao += 2

sedentario = input("Está há mais de 3 meses sem praticar atividade física (sim/não): ").lower()
if sedentario == "sim":
    pontuacao -= 2

disposicao = input("Notou indisposição para praticar atividades físicas por motivos de preocupação, estresse ou tristeza (sim/não): ").lower()
if disposicao == "sim":
    pontuacao -= 2

print("\n ====== HÁBITOS DE RISCO ======")

tabagismo = input("Tem o costume de fumar (sim/não): ").lower()
if tabagismo == "sim":
    pontuacao -= 2

alcool = input("Consome bebida alcóolica com frequência (sim/não): ").lower()
if alcool == "sim":
    pontuacao -= 2

doce = input("Consome alimentos doces com freqência (sim/não): ").lower()
if doce == "sim":
    pontuacao -= 2

sal = input("Tem o costume de adicionar mais sal em sua comida (sim/não): ").lower()
if sal == "sim":
    pontuacao -= 2


print("\n ====== INDICADORES CORPORAIS ======")

peso = float(input("Informe seu peso atual (kg): "))
altura = float(input("Informe sua altura (m): "))
if altura > 0:
    imc = peso / (altura**2)
    print(f"\n Seu IMC calculado é: {imc:.2f}")
    if imc< 18.5:
      prontuacao -= 1
      print("Classificação: VOCÊ ESTÁ ABAIXO DO PESO!!")

    elif 18.5 <= imc <= 24.9:
        pontuacao += 2
        print("Classificação: SEU PESO ESTÁ NA FAIXA ADEQUADA!!")
    elif 25 <= imc <= 29.9:
        pontuacao -= 1
        Print("Classificação: ATENÇÃO, VOCÊ ESTÁ NA FAIXA DE SOBREPESO!!")
    else:
        pontuacao -= 2
        Print("Classificação: ATENÇÃO, VOCÊ ESTÁ NA FAIXA DE OBESIDADE!!")
else:
    print("ALTURA INVÁLIDA!")

cintura = float(input("Informe a medida de sua circunferência abdominal (cm): "))
if sexo == "masculino" and cintura < 94:
    pontuacao += 2
elif sexo == "feminino" and cintura < 80:
    pontuacao += 2

print("\n ====== OCORRÊNCIA DE DORES E INCÔMODOS ======")

dores = input("Costuma sentir dores físicas frequentemente (sim/não): ").lower()
if dores == "sim":
    pontuacao -= 2

incomodos = input("Costuma sentir dores de cabeça frequentemente (sim/não): ").lower()
if incomodos == "sim":
    pontuacao -= 2

print("\n ====== RESULTADOS ======")

print(f"SUA PONTUAÇÃO GERAL: {pontuacao}")
if pontuacao >= 20:
    classificacao = "SAÚDE EQUILIBRADA!"
elif pontuacao >= 12:
    classificacao = "ATENÇÃO: MELHORE SEUS HÁBITOS"
else:
    classificacao = "PONTO CRÍTICO: SUA SÁUDE ESTÁ EM ZONA DE RISCO!"

print(f"CLASSIFICAÇÃO: {classificacao}")

print("\n ====== RELATÓRIO ======")

if idade >= 60:
  print("IDADE AVANÇADA: MANTER ACOMPANHAMENTO REGULAR JUNTO A UM PROFISSIONAL DE SAÚDE")
elif idade >= 30:
  print("ATENÇÃO: REFORCE HÁBITOS SAUDÁVEIS COMO:")
  print("ALIMENTAÇÃO EQUILIBRADA E ATIVIDADES FÍSICAS REGULARES")
  print("TENHA UM TEMPO PARA DESCANSO E MANTENHA SUAS CONSULTAS E EXAMES MÉDICOS ATUALIZADOS!")

if horas_sono < 5 and idade >= 30:
  print("SONO INSUFICIENTE: NA SUA FAIXA ETÁRIA O RISCO PARA SUA SAÚDE É MAIOR!")

if frequencia == "nao" and idade >= 30:
  print("A FALTA DE ATIVIDADE FÍSICA AUMENTA O RISCO DE DOENÇAS COM O AVANÇO DA IDADE!!")

if dores == "sim" and incomodos == "sim" and idade >= 30:
  print("A FREQUÊNCIA DE DORES E OUTROS SINTOMAS NO CORPO NA SUA FAIXA ETÁRIA REQUER CUIDADO E ATENÇÃO!")
  print("PROCURE IMEDIATAMENTE AUXÍLIO PROFISSIONAL!")

print("\n ====== ALERTAS ======")

if fruta == "não":
    print("BAIXO CONSUMO DE FRUTAS!")
    print("FRUTAS AJUDAM NO FORTALECIMENTO DO ORGANISMO!")
    print("PROCURE CONSUMIR PELO MENOS DUAS FRUTAS AO LONGO DA SEMANA!")
    print("=============================================================")

if lanche == "sim":
    print("NÃO É ACONSELHÁVEL SUBSTITUIR REFEIÇÃO POR LANCHE!")
    print("=============================================================")

if bebida == "sim" or comida == "sim":
    print("CUIDADO COM OS ALIMENTOS INDUSTRIALIZADOS!")
    print("O EXCESSO TRAZ SÉRIOS RISCOS À SAÚDE A MÉDIO E LONGO PRAZO!!")
    print("=============================================================")

if horas_sono < 5:
    print("SONO INSUFICIENTE TRAZ RISCOS PARA SUA SAÚDE!")
    print("SONO INSUFICIENTE AFETA O DESEMPENHO CRIATIVO E A PRODUTIVIDADE")
    print("=============================================================")

if tabagismo == "sim" or alcool == "sim" or doce == "sim" or sal == "sim":
    print("EVITE QUE SEUS HÁBITOS SEJAM A PORTA PRINCIPAL PARA DOENÇAS!!")
    print("=============================================================")

if classificacao == "SAÚDE EQUILIBRADA":
  print("PARABÉNS! CONTINUE MANTENDO O FOCO!")
  print("BONS HÁBITOS E CUIDADOS NECESSÁRIOS FAZEM TODA A DIFERENÇA!")
  print("=============================================================")

elif classificacao == "ATENÇÃO: MELHORE SEUS HÁBITOS!!":
  print("ALGUNS HÁBITOS PRECISAM MELHORAR E ALGUNS CUIDADOS PRECISAM SER REDOBRADOS!!")
  print("=============================================================")

else:
  print("SUA SAÚDE ESTÁ EM ZONA DE ALTO RISCO!")
  print("PROCURE AUXÍLIO PROFISSIONAL PARA SANAR DÚVIDAS")
  print("SAÚDE SEMPRE EM PRIMEIRO LUGAR!!")
  print("=============================================================")

print("ESTA AVALIAÇÃO É APENAS UM EXERCÍCIO EDUCATIVO E CONSCIENTIZADOR!!")
print("VISA UNICAMENTE EDUCAR E SENSIBILIZAR PARA HÁBITOS MELHORES E CUIDADOS COM A SAÚDE!!")
print("PORTANTO, ESTA AVALIAÇÃO NÃO SUBSTITUI CONSULTAS, EXAMES OU AS RECOMENDAÇÕES MÉDICAS!!")
print("=============================================================")

```
