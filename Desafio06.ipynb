```python
#SISTEMA DE TOMADA DE DECISÃO#

print("\n #====== SISTEMA DE TOMADA DECISÃO ======#")

pontuacao = 0

#DECISÃO#

decisao = input("Informe a Decisão que Você deseja avaliar: ")

beneficio = input("O benefício é (alto/medio/baixo): ").lower()
if beneficio == "alto":
  pontuacao += 2
elif beneficio == "medio":
  pontuacao += 1

#CUSTO BENEFÍCIO#

custo = input("O custo financeiro é (alto/médio/baixo): ").lower()
if custo == "alto":
  pontuacao -= 2
elif custo == "médio":
  pontuacao -= 1

tempo = input("Informe o tempo necessário para a realização (alto/médio/baixo): ").lower()
if tempo == "alto":
  pontuacao -= 2
elif tempo == "médio":
  pontuacao -= 1

#CONSEQUÊNCIA#
risco = input("Informe o nível de risco da sua decisão (alto/médio/baixo): ").lower()
if risco == "alto":
  pontuacao -= 2
elif risco == "médio":
  pontuacao -= 1

consequencia = input("Essa decisão melhora sua vida (sim/não): ").lower()
if consequencia == "sim":
  pontuacao += 2

objetivo = input("Essa decisão contribui para alcançar o seu objetivo principal (totalmente/parcialmente/não): ")
if objetivo == "totalmente":
  pontuacao += 2
elif objetivo == "parcialmente":
  pontuacao += 1

#RESULTADO#

print("\n #===== RESULTADO ======#")

print(f"Decisão Analisada: {decisao}")
print(f"Pontação Alcançada: {pontuacao}")
if pontuacao >= 4:
  print("DECISÃO FAVORÁVEL")
elif pontuacao >= 1:
  print("AVALIE MELHOR O CENÁRIO")
  print("PONDERE SOBRE O QUE É NECESSÁRIO ACRESCENTAR, DESCARTAR OU AJUSTAR")
else:
  print("DECISÃO ARRISCADA!")
  print("REVISE O PLANO")
  print("CONSIDERE AGUARDAR UM MOMENTO QUE LHE FAVOREÇA")
```
